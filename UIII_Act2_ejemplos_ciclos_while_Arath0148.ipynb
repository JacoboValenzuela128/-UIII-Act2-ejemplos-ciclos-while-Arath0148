#### Ejemplos de Ciclos while

```python
# 1. Suma interactiva con centinela
# Este programa suma números ingresados por el usuario hasta que este decida detenerse ingresando el número 0
total = 0
numero = 1
while numero != 0:
    numero = int(input("Ingresa un número para sumar (0 para salir): "))
    total = total + numero
print(f"La suma total es: {total}")
# Línea 1: Inicializa la variable total en 0 para acumular la suma.
# Línea 2: Inicializa numero en 1 para asegurar que la primera condición del while sea verdadera.
# Línea 3: El ciclo se repite mientras el número ingresado sea distinto de 0.
# Línea 4: Solicita al usuario un dato y lo convierte a entero usando int().
# Línea 5: Suma el valor ingresado al acumulador total.
# Línea 6: Muestra el resultado final una vez que el ciclo termina.
```

```python
# 2. Adivinar el número secreto
# El programa solicita números hasta que el usuario logre igualar el valor almacenado en la variable "secreta"
secreto = 7
intento = 0
while intento != secreto:
    intento = int(input("Adivina el número secreto (1 al 10): "))
print("¡Felicidades! Adivinaste el número.")
# Línea 1: Define el número que el usuario debe encontrar.
# Línea 2: Inicializa la variable intento con un valor que no sea el secreto.
# Línea 3: Evalúa si el intento es diferente al número secreto para continuar pidiendo datos.
# Línea 4: Pide al usuario que ingrese su apuesta numérica.
# Línea 5: Imprime un mensaje de éxito cuando la condición del while se vuelve falsa (cuando son iguales).
```

```python
# 3. Validador de rango (Meses del año)
# Este ejemplo obliga al usuario a ingresar un dato válido (entre 1 y 12) antes de permitirle continuar
mes = 0
while mes < 1 or mes > 12:
    mes = int(input("Ingresa un número de mes válido (1-12): "))
print(f"Has seleccionado el mes número: {mes}")
# Línea 1: Se inicializa mes en 0 para que entre al ciclo la primera vez.
# Línea 2: El ciclo se ejecuta mientras el valor esté fuera del rango 1-12 (menor a 1 o mayor a 12).
# Línea 3: Solicita la entrada del dato al usuario.
# Línea 4: Muestra el resultado únicamente cuando el dato es correcto.
```

```python
# 4. Acumulador de ahorros interactivo
# Permite al usuario ingresar montos de dinero para ahorrar hasta llegar a una meta específica.
meta = 100
ahorro = 0
while ahorro < meta:
    deposito = int(input("¿Cuánto dinero deseas depositar hoy?: "))
    ahorro = ahorro + deposito
print(f"¡Meta alcanzada! Tienes ahorrados: {ahorro}")
# Línea 1: Establece el objetivo económico a alcanzar.
# Línea 2: Inicializa el contador de ahorro en 0.
# Línea 3: El ciclo continúa mientras el ahorro actual sea menor que la meta.
# Línea 4: Pide al usuario el monto que desea agregar a su ahorro.
# Línea 5: Actualiza el valor de ahorro sumándole el nuevo depósito.
# Línea 6: Informa al usuario que ha completado su meta.
```

```python
# 5. Control de acceso simple
# El ciclo solicita una clave de texto y no termina hasta que el usuario escriba la palabra correcta.
clave_correcta = "python123"
entrada = ""
while entrada != clave_correcta:
    entrada = input("Ingresa la contraseña de acceso: ")
print("Acceso concedido al sistema.")
# Línea 1: Define la cadena de texto que actúa como contraseña.
# Línea 2: Inicializa entrada como una cadena vacía.
# Línea 3: Compara la entrada del usuario con la clave; se repite mientras sean diferentes.
# Línea 4: Captura el texto ingresado por el usuario a través del teclado.
# Línea 5: Muestra el mensaje de éxito tras salir del bucle.

print("Programa(s) realizado(s) por Arath Valenzuela NL52 Nc0148")
```
